4. Дан 1D массив, удалите все элементы которые входят в диапазон `[3, 8]` inplace

````python
arr = [0, 4, 4, 2, 1, 4, 8, 7, 10, 15]

np_arr = np.array([0, 4, 4, 2, 1, 4, 8, 7, 10, 15])

mask = (np_arr < 3) | (np_arr > 8)
count = np.count_nonzero(mask)

np_arr[:count] = np_arr[mask]
np_arr = np_arr[:count]

print(np_arr)
````

````python
import pandas as pd
````

````python
# Load dataset here
df = pd.read_csv('russian-housing-market.csv')
df.head()
````

1. Посчитайте количество записей в датасете с отсутствующей информацией о жилой площади жилого помещения

````python
null_count = df['life_sq'].isnull().sum()
print(null_count)
````

2. Предложите способ заменить отсутствующие значения из пункта 1 и реализуйте его. Способ нужно аргументировать(сказать почему это достаточно хорошее приближение)

````python
ratio = (df['life_sq'] / df['full_sq']).median()
appr_life_sq = df['full_sq'] * ratio
df['life_sq'] = df['life_sq'].fillna(appr_life_sq)

print(df['life_sq'].isnull().sum(), '\n')
print(df)
````

Пояснение к заданию 2

Жилая и общая площади имеют сильную зависисимость. Если заполнить пропуски медианой или средним по столбцу, то данные сильно исказятся: у многокомнатной квартиры может оказаться та же площадь, что и у студии. Использование пропорции позволяет сохранить реалистичный масштаб.

9. Проанализируйте есть ли разница в цене между квартирами на 1 и 9 этажах

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

````python
import matplotlib.pyplot as plt

df['timestamp'] = pd.to_datetime(df['timestamp'])
````

1. Как менялась средняя цена квартиры в Нагатинском Затоне(Nagatinski Zaton) с течением времени

````python
NZ_df = df[df['sub_area'] == 'Nagatinskij Zaton'].copy()
NZ_df['price_mln'] = NZ_df['price_doc'] / 1e6

monthly_stats = NZ_df.set_index('timestamp').resample('ME')['price_mln'].mean().dropna()

plt.figure(figsize=(12, 6))

plt.plot(monthly_stats.index, monthly_stats, marker='o', markersize=4)

plt.title('Динамика цен на квартиры в Нагатинском Затоне (помесячно)', fontsize=14)
plt.xlabel('Дата', fontsize=12)
plt.ylabel('Цена (млн руб.)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
````

2. Постройте гистограмму распределения цен на квартиры с общей площадью более 50 кв.метров

````python
import matplotlib.pyplot as plt

filtered = df[df['full_sq'] > 50]['price_doc'] / 1e6

plt.figure(figsize=(10, 5))
plt.hist(filtered, bins=80, edgecolor='white', alpha=0.7)

plt.title('Распределение цен на квартиры площадью более 50 кв. м')
plt.xlabel('Цена (млн руб.)')
plt.ylabel('Количество квартир')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
````

3. Визуализируйте количество проданных квартир в разных районах

````python
areas = df['sub_area'].value_counts().head(20)

plt.figure(figsize=(12, 6))
areas.sort_values().plot(kind='barh')

plt.title('Районы по количеству проданных квартир')
plt.xlabel('Количество проданных квартир')
plt.ylabel('Район')
plt.grid(axis='x', linestyle=':', alpha=0.6)

plt.show()
````

4. Постройте box plot отображающий распределение полных площадей квартир в Басманном районе

````python
basmannoe_sq = df[df['sub_area'] == 'Basmannoe']['full_sq']

plt.figure(figsize=(4, 6))
plt.boxplot(basmannoe_sq.dropna())

plt.title('Распределение площади в Басманном районе')
plt.ylabel('Полная площадь квм')
plt.xlabel('Басманное')
plt.grid(True, linestyle=':', alpha=0.6)

plt.show()
````

5. Визуализируйте зависимость средней стоимости квартир в зависимости от этажа

````python
avg_price = df.groupby('floor')['price_doc'].mean() / 1e6

plt.figure(figsize=(12, 6))
avg_price.plot(kind='bar')

plt.title('Зависимость средней стоимости квартир от этажа')
plt.xlabel('Этаж')
plt.ylabel('Средняя стоимость')
plt.grid(axis='y', linestyle='-', alpha=0.6)
plt.show()
````

6. Покажите как меняется средняя площадь квартир в зависимости от района

````python
avg_area = df.groupby('sub_area')['full_sq'].mean().sort_values(ascending=False).head(20)

print(avg_area)

plt.figure(figsize=(12, 6))
avg_area.plot(kind='bar', edgecolor='white')

plt.title('Средняя площадь квартир в зависимости от района')
plt.xlabel('Район')
plt.ylabel('Средняя площадь')

plt.grid(axis='y', linestyle='-', alpha=0.6)
plt.tight_layout()
plt.show()
````

7. Постройте график изменения медианной цены на квартиру менее 50 кв. метров за период с 2011 до 2014 года*

\* Задание на доп. балл

````python
df['year'] = df['timestamp'].dt.year

n_filtered = df[(df['full_sq'] < 50) & (df['year'].between(2011, 2014))].copy()
n_filtered['month'] = n_filtered['timestamp'].dt.to_period('M')
median = n_filtered.groupby('month')['price_doc'].median() / 1e6
median.index = median.index.to_timestamp()

plt.Figure(figsize=(12, 6))
plt.plot(median.index, median.values, marker='o', markersize=4, color='royalblue')

plt.title('Динамика медианной цены квартир')
plt.xlabel('Год')
plt.ylabel('Медианная цена продажи')
plt.gcf().autofmt_xdate()
plt.grid(True, linestyle='-', alpha=0.6)

plt.show()
````
